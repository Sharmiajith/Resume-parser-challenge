```python
import pdfplumber
import re
import spacy
import json

# Load spaCy NLP model
nlp = spacy.load("en_core_web_sm")

# Extract text from PDF
def extract_text_from_pdf(pdf_path):
    text = ''
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            text += page.extract_text() + '\n'
    return text

# Extract email and phone using regex
def extract_contact_info(text):
    email = re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", text)
    phone = re.findall(r"\+?\d[\d\s-]{8,}\d", text)
    return email[0] if email else None, phone[0] if phone else None

# Extract name using NER (first PERSON entity)
def extract_name(text):
    doc = nlp(text)
    for ent in doc.ents:
        if ent.label_ == "PERSON":
            return ent.text
    return None

# Extract skills using a keyword match (simplified)
def extract_skills(text, skill_set):
    found_skills = [skill for skill in skill_set if skill.lower() in text.lower()]
    return list(set(found_skills))

# Dummy skill set for matching
common_skills = [
    "Python", "Java", "C++", "Power BI", "Excel", "Machine Learning",
    "Project Management", "Civil 3D", "AutoCAD", "Revit", "PyTorch"
]
#extract education
def extract_education(text):
  edu_keywords = ['Bachelor', 'Master', 'B.Tech', 'M.Tech', 'B.Sc', 'M.Sc', 'B.E', 'MBA', 'PhD']
  education_lines = []
  for line in text.split('\n'):
    if any(keyword in line for keyword in edu_keywords):
     education_lines.append(line.strip())
    return education_lines

common_education = [ "Bachelor", "Master", "B.Tech", "M.Tech", "B.Sc",
                    "M.Sc", "B.E", "M.E", "MBA", "PhD", "High School", "Diploma", "Computer Science",
                    "Engineering", "Information Technology", "Electronics", "Mechanical", "Civil", "Data Science"
]

# extract certificates
def extract_certificates(text):
  cert_keywords = ['certificate', 'certified', 'certification', 'course']
  certificate_lines = []
  for line in text.split('\n'):
    if any(keyword.lower() in line.lower()
     for keyword in cert_keywords):
      certificate_lines.append(line.strip())
    return certificate_lines

#linked in
def extract_linkedin(text):
  match = re.search(r"(https?://(www.)?linkedin.com/in/[A-Za-z0-9-_/]+)", text)
  if not match:
    return None
    return match.group(0)

# EXTRACT experience
def extract_experience(text, experience_keywords):
   experience_lines = []
   for line in text.split('\n'):
    if any(keyword.lower() in line.lower()
    for keyword in experience_keywords):
       experience_lines.append(line.strip())
       return list(set(experience_lines))

common_experience_keywords = [ "intern", "internship", "experience", "worked",
                              "developed", "engineer", "analyst", "consultant",
                               "manager", "supervised", "project", "responsible for", "team lead",
                               "software developer", "data scientist", "executed", "built"
]

# Main function to parse a resume
def parse_resume(pdf_path):
    text = extract_text_from_pdf(pdf_path)
    name = extract_name(text)
    email, phone = extract_contact_info(text)
    skills = extract_skills(text, common_skills)
    education=extract_education(text)
    certificates=extract_certificates(text)
    linkedin=extract_linkedin(text)
    experience = extract_experience(text, common_experience_keywords)


    result = {
    "name": name,
    "email": email,
    "phone": phone,
    "skills": skills,
    "education": education,
    "certificates": certificates,
    "linkedin": linkedin,
    "experience":experience
}


    return result

```
